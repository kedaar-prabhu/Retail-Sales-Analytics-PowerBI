```python
import pandas as pd
import numpy as np

df=pd.read_csv('/content/adult_clean.csv')
df.head()
```

```python
df.drop('Unnamed: 0',axis=1,inplace=True)
```

```python
df.head()
```

```python
df.boxplot(column='age')  #finding and removing outlier
```

```python
Q1=df['age'].quantile(0.25)
Q3=df['age'].quantile(0.75)
IQR=Q3-Q1
upper_limit=Q3+(1.5*IQR)
lower_limit=Q1-(1.5*IQR)

median=(df['age']).median()
df['age']=np.where(df['age']>upper_limit,median,df['age'])
df['age']=np.where(df['age']<lower_limit,median,df['age'])

df.boxplot(column='age')
```

```python
#analysis
#filter
df[  df['age']>50 ]
```

```python
df[ ( df['age']>50) & (df['sex']=='Male')]
```

```python
#sort
df[ ( df['age']>50) & (df['sex']=='Male')].sort_values(by ='age') #(ascending)
```

```python
df[df['hours-per-week']>40].sort_values(by='hours-per-week',ascending=False)
```

Complete the project by finding the factors affecting the income of individuals.

```python
import pandas as pd
import numpy as np

# Analyzing the core factors driving Performance Score
numeric_factors = ['age', 'hours-per-week', 'capital-gain', 'capital-loss']
correlation_matrix = df[numeric_factors].corr()

("--- Correlation of Numerical Factors with Income Factors ---")
(correlation_matrix.unstack().sort_values(ascending=False))

("\n--- Categorical Factors: Work Mode Impact ---")
(df.groupby(['workclass', 'income'])['hours-per-week'].mean().unstack())
```

### Impact of Numerical Factors on Income

```python
(df.groupby('income')[['age', 'hours-per-week', 'capital-gain', 'capital-loss']].mean().round(2))
```

### Visualizing Categorical Factors vs. Income

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Education vs. Income
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='education', hue='income', palette='viridis')
plt.title('Income Distribution by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Marital Status vs. Income
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='marital-status', hue='income', palette='magma')
plt.title('Income Distribution by Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Occupation vs. Income
plt.figure(figsize=(14, 7))
sns.countplot(data=df, x='occupation', hue='income', palette='cividis')
plt.title('Income Distribution by Occupation')
plt.xlabel('Occupation')
plt.ylabel('Count')
plt.xticks(rotation=60, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Sex vs. Income
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='sex', hue='income', palette='plasma')
plt.title('Income Distribution by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
